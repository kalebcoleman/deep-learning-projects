In this notebook we will learn a fully-connected NN for binary classification. Specifically we will use the same Heart Disease data that we saw for Logistic Regression. The first code block will load that data, subset the columns to the 3 we are interested in, and then standardize the data.

```python
# imports
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# Load the heart disease CSV file into a DataFrame
df_heart_disease = pd.read_csv('https://raw.githubusercontent.com/benjaminmlucas/MAT499/refs/heads/main/module_1/heart_disease.csv')

# Define the predictor variables (X) and target variable (y) from the subset
X = df_heart_disease[['chol', 'age', 'trestbps']]
y = df_heart_disease['target']

# Initialize the StandardScaler
scaler = StandardScaler()

# Fit the scaler to the data and transform it
X_scaled = scaler.fit_transform(X)

# Convert the scaled array back to a DataFrame for easier viewing
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)

# check what it looks like
display(X_scaled.head())
```

Now we want to define the model. The following code block is blank for you to define a model in Torch. Attempt to write a class for a fully-connected NN with one hidden layer with 8 neurons.

```python
# here are some imports that'll be useful
import torch
import torch.nn as nn
import torch.optim as optim
```

Now create the training loop where you learn the parameters with gradient descent.

```python
class BinaryClassifier(nn.Module):
    def __init__(self, input_dim):
        super(BinaryClassifier, self).__init__()
        # Define the hidden layer with 8 neurons
        self.linear1 = nn.Linear(input_dim, 8)
        self.relu = nn.ReLU()
        # Define the output layer
        self.linear2 = nn.Linear(8, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        # Pass through the hidden layer and apply ReLU
        x = self.linear1(x)
        x = self.relu(x)
        # Pass through the output layer and apply Sigmoid for binary classification probability
        x = self.linear2(x)
        x = self.sigmoid(x)
        return x
```

```python
X_torch = torch.tensor(X_scaled.values, dtype=torch.float32)
Y_torch = torch.tensor(y.values, dtype=torch.float32).view(-1, 1)

model = BinaryClassifier(input_dim=X_torch.shape[1])

# Print random intitial values for weights and bia
with torch.no_grad():
    print(f'Initial Values: Weights = {model.linear1.weight.numpy()[0].round(decimals=3)}, \
Bias = {model.linear1.bias.numpy().round(decimals=3)}')

# Define our loss function and optimizer Binary Cross Entrop Loss
criterion = nn.BCELoss()
# Sthocastic gradient descent 
optimizer = optim.SGD(model.parameters(), lr=0.05)

epochs = 100000

print()
print('Starting Training with gradient descent')
print()
for epoch in range(epochs):
    
    # pass entire dataset X at once to the model and get predictions 
    y_pred = model(X_torch)

    # calculate cost/loss
    loss = criterion(y_pred, Y_torch)

    # update model parameters
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (epoch+1) % 1000 == 0:
      with torch.no_grad():
        predicted_labels = (y_pred > 0.5).float()
        accuracy = (predicted_labels == Y_torch).float().mean()
        print(f'Epoch {epoch+1}: Loss = {loss.item():.5f}, Accuracy = {accuracy.item():.2f}')
print()
print('Training Complete')
print()

#print the final learned values
with torch.no_grad():
  print(f'Final Trained Values: Weights = {model.linear1.weight.numpy()[0].round(decimals=3)}, \
    Bias = {model.linear1.bias.numpy().round(decimals=3)}')
```
